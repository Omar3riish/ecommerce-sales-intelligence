```python
df_raw = df.copy()
df = df.copy()
```

```python
missing_summary = (
    df.isna()
      .sum()
      .to_frame("Missing_Count")
      .assign(Missing_Percentage=lambda x: (x["Missing_Count"] / len(df) * 100).round(2))
      .sort_values("Missing_Count", ascending=False)
)
display(missing_summary)
```

```python


display(df[df["CouponCode"].isna()].head(10))
```

```python
print("Exact duplicate rows:", df.duplicated().sum())
print("Duplicate OrderID values:", df["OrderID"].duplicated().sum())
print("Unique OrderIDs:", df["OrderID"].nunique(dropna=True))
print("Unique CustomerIDs:", df["CustomerID"].nunique(dropna=True))
print("Missing OrderIDs:", df["OrderID"].isna().sum())
print("Missing CustomerIDs:", df["CustomerID"].isna().sum())
```

```python
customer_order_counts = df["CustomerID"].value_counts(dropna=True)
repeated_customers = customer_order_counts[customer_order_counts > 1]

print("Customers with multiple orders:", len(repeated_customers))
print("Orders belonging to repeat customers:", int(repeated_customers.sum()))

display(repeated_customers.rename("Order_Count").to_frame().head(20))
```

```python
repeat_customer_orders = (
    df[df["CustomerID"].isin(repeated_customers.index)]
      .sort_values(["CustomerID", "Date", "OrderID"])
)
display(repeat_customer_orders)
```

**Decision:** Do not remove repeated `CustomerID` values. Investigate repeated `OrderID` values or exact duplicate rows individually before deciding whether any record should be removed.

```python
if "Date" in df.columns:
    date_text = df["Date"].astype("string").str.strip()
    expected_format = date_text.str.fullmatch(r"\d{4}-\d{2}-\d{2}", na=False)

    print("Values not matching YYYY-MM-DD:", int((date_text.notna() & ~expected_format).sum()))
    display(df.loc[date_text.notna() & ~expected_format, ["Date"]].head(20))
```

```python
df["Date"] = pd.to_datetime(df["Date"], format="%Y-%m-%d", errors="coerce")

print("Date dtype:", df["Date"].dtype)
```

```python
numeric_cols = ["Quantity", "UnitPrice", "ItemsInCart", "TotalPrice"]

for col in numeric_cols:
    df[col] = pd.to_numeric(df[col], errors="coerce")

display(df[numeric_cols].describe().T)
display(df[numeric_cols].isna().sum().to_frame("Missing_Count"))
```

```python
calculated_total = df["Quantity"] * df["UnitPrice"]

valid_comparison = df["TotalPrice"].notna() & calculated_total.notna()
price_mismatch = valid_comparison & ~np.isclose(
    df["TotalPrice"],
    calculated_total,
    atol=0.01,
    rtol=0
)


display(
    df.loc[price_mismatch, ["OrderID", "Quantity", "UnitPrice", "TotalPrice"]]
      .assign(CalculatedTotal=calculated_total[price_mismatch])
      .head(20)
)
```

```python
categorical_cols = df.select_dtypes(include=["object", "string"]).columns

for col in categorical_cols:
    print(f"\n--- {col} ---")
    display(df[col].value_counts(dropna=False).to_frame("Count"))
```

```python
df_numeric_long = df[numeric_cols].melt(var_name="Feature", value_name="Value")

fig = px.box(
    df_numeric_long,
    x="Feature",
    y="Value",
    points="outliers",
    title="Numerical Features — Box Plots"
)
fig.update_layout(template="plotly_white", height=550)
fig.show()
```

```python
df_numeric_long = df[numeric_cols].melt(var_name="Feature", value_name="Value")

fig = px.box(
    df_numeric_long,
    x="Feature",
    y="Value",
    facet_col="Feature",
    facet_col_wrap=2,
    points="outliers",
    title="Numerical Features — Box Plots (Independent Scales)"
)
fig.update_yaxes(matches=None)
fig.update_layout(template="plotly_white", height=700, showlegend=False)
fig.show()
```

```python
Q1 = df["TotalPrice"].quantile(0.25)
Q3 = df["TotalPrice"].quantile(0.75)
IQR = Q3 - Q1

lower_fence = Q1 - 1.5 * IQR
upper_fence = Q3 + 1.5 * IQR

totalprice_outliers = df[
    (df["TotalPrice"] < lower_fence) |
    (df["TotalPrice"] > upper_fence)
]

print(f"Lower fence: {lower_fence:.2f}")
print(f"Upper fence: {upper_fence:.2f}")
print("Potential TotalPrice outliers:", len(totalprice_outliers))

display(
    totalprice_outliers[
        ["OrderID", "Date", "CustomerID", "Product", "Quantity", "UnitPrice", "TotalPrice"]
    ].sort_values("TotalPrice", ascending=False)
)
```

```python
df["YearMonth"] = df["Date"].dt.to_period("M").astype("string")
df["OrderMonth"] = df["Date"].dt.to_period("M").dt.to_timestamp()
df["DayOfWeek"] = df["Date"].dt.day_name()
df["OrderYear"] = df["Date"].dt.year
```

```python
orders_by_product = (
    df.groupby("Product", dropna=False)
      .agg(Order_Count=("OrderID", "nunique"), Units_Sold=("Quantity", "sum"), Revenue=("TotalPrice", "sum"))
      .reset_index()
      .sort_values("Order_Count", ascending=False)
)
display(orders_by_product)

fig = px.bar(
    orders_by_product, x="Product", y="Order_Count", color="Product",
    title="Number of Orders by Product", text="Order_Count"
)
fig.update_layout(template="plotly_white", showlegend=False)
fig.show()
```

```python
revenue_by_product = orders_by_product.sort_values("Revenue", ascending=False)

fig = px.bar(
    revenue_by_product, x="Product", y="Revenue", color="Product",
    title="Revenue by Product", text_auto=".2s"
)
fig.update_layout(template="plotly_white", showlegend=False, yaxis_title="Revenue")
fig.show()
```

```python
fig = px.bar(
    orders_by_product.sort_values("Units_Sold", ascending=False),
    x="Product", y="Units_Sold", color="Product",
    title="Units Sold by Product", text="Units_Sold"
)
fig.update_layout(template="plotly_white", showlegend=False)
fig.show()
```

```python
monthly_sales = (
    df.dropna(subset=["OrderMonth"])
      .groupby("OrderMonth", as_index=False)
      .agg(Revenue=("TotalPrice", "sum"), Orders=("OrderID", "nunique"))
      .sort_values("OrderMonth")
)
display(monthly_sales)

fig = px.line(
    monthly_sales, x="OrderMonth", y="Revenue", markers=True,
    title="Monthly Revenue Trend"
)
fig.update_layout(template="plotly_white", xaxis_title="Month", yaxis_title="Revenue")
fig.show()
```

```python
fig = px.bar(
    monthly_sales, x="OrderMonth", y="Orders",
    title="Number of Orders per Month", text="Orders"
)
fig.update_layout(template="plotly_white", xaxis_title="Month", yaxis_title="Orders")
fig.show()
```

```python
import plotly.express as px

status_counts = (
    df["OrderStatus"].value_counts(dropna=False)
      .rename_axis("OrderStatus")
      .reset_index(name="Order_Count")
)
display(status_counts)

fig = px.pie(
    status_counts,
    names="OrderStatus",
    values="Order_Count",
    title="Orders by Status",
    color="OrderStatus"
)
fig.update_traces(
    textposition="inside",
    textinfo="percent+label",
    hovertemplate="<b>%{label}</b><br>Orders: %{value}<br>Share: %{percent}<extra></extra>"
)
fig.update_layout(template="plotly_white", showlegend=True)
fig.show()
```

```python
import plotly.express as px

payment_summary = (
    df.groupby("PaymentMethod", dropna=False)
      .agg(Orders=("OrderID", "nunique"), Revenue=("TotalPrice", "sum"))
      .reset_index()
      .sort_values("Revenue", ascending=False)
)
display(payment_summary)

fig = px.pie(
    payment_summary,
    names="PaymentMethod",
    values="Orders",
    title="Orders by Payment Method",
    color="PaymentMethod"
)
fig.update_traces(
    textposition="inside",
    textinfo="percent+label",
    hovertemplate="<b>%{label}</b><br>Orders: %{value}<br>Share: %{percent}<extra></extra>"
)
fig.update_layout(template="plotly_white", showlegend=True)
fig.show()
```

```python
fig = px.bar(
    payment_summary, x="PaymentMethod", y="Revenue", color="PaymentMethod",
    title="Revenue by Payment Method", text_auto=".2s"
)
fig.update_layout(template="plotly_white", showlegend=False)
fig.show()
```

```python
orders_per_customer = (
    df.groupby("CustomerID", dropna=True)["OrderID"]
      .nunique()
      .rename("Order_Count")
      .reset_index()
)

display(orders_per_customer["Order_Count"].describe())

fig = px.histogram(
    orders_per_customer, x="Order_Count", nbins=2,
    title="Distribution of Orders per Customer"
)
fig.update_layout(template="plotly_white", xaxis_title="Number of Orders", yaxis_title="Number of Customers")
fig.show()
```

```python
product_status = (
    df.groupby(["Product", "OrderStatus"], dropna=False)
      .size()
      .reset_index(name="Order_Count")
)

fig = px.bar(
    product_status, x="Product", y="Order_Count", color="OrderStatus",
    barmode="stack", title="Order Status by Product"
)
fig.update_layout(template="plotly_white")
fig.show()
```

```python
weekday_order = (
    df.dropna(subset=["DayOfWeek"])
      .groupby("DayOfWeek")
      .agg(Orders=("OrderID", "nunique"), Revenue=("TotalPrice", "sum"))
      .reindex(["Monday", "Tuesday", "Wednesday", "Thursday", "Friday", "Saturday", "Sunday"])
      .reset_index()
)
display(weekday_order)

fig = px.bar(
    weekday_order, x="DayOfWeek", y="Orders",
    title="Orders by Day of Week", text="Orders"
)
fig.update_layout(template="plotly_white")
fig.show()
```
